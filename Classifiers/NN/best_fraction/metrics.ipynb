```python
import pandas as pd
from sklearn.metrics import recall_score, accuracy_score, precision_score, f1_score, confusion_matrix
from sklearn import metrics
```

```python
table_test_1 = pd.read_excel("predictions_test_1.xlsx", index_col="Unnamed: 0")

table_test_1.head()
```

```python
table_test_2 = pd.read_excel("predictions_test_2.xlsx", index_col="Unnamed: 0")

table_test_2.head()
```

```python
def score_printer(header, col_name, table):
    confusion = confusion_matrix(table['label'], table[col_name])
    print(f"Multibranch score - {header}")
    # print(f"Precision: {round(precision_score(table['label'], table[col_name], average='binary'), 2)}")
    # print(f"Recall: {round(recall_score(table['label'], table[col_name], average='binary'), 2)}")
    # print(f"F1 Score: {round(f1_score(table['label'], table[col_name], average='binary'), 2)}")
    # print(f"Accuracy: {round(accuracy_score(table['label'], table[col_name]), 2)}")

    total=sum(sum(confusion))
    Accuracy = (confusion[0,0]+confusion[1,1])/total
    Specificity = confusion[0,0]/(confusion[0,0]+confusion[0,1])
    Sensitivity = confusion[1,1]/(confusion[1,0]+confusion[1,1])
    Precision = confusion[1,1]/(confusion[0,1]+confusion[1,1])
    Recall = confusion[1,1]/(confusion[1,0]+confusion[1,1]) 
    F1 = 2 * (Precision * Recall) / (Precision + Recall)
    print(f"Precision: {round(Precision, 2)}")
    print(f"Recall: {round(Recall, 2)}")
    print(f"F1 Score: {round(F1, 2)}")
    print(f"Accuracy: {round(Accuracy, 2)}")
    print(f"Specificity: {round(Specificity, 2)}")
    print(f"Sensitivity: {round(Sensitivity, 2)}")
```

```python
col_name = 'Multibranch_predict - train 1'
header = "train 1, test 1"
score_printer(header, col_name, table_test_1)
```

```python
for table, table_name in [(table_test_1, "test_1"), (table_test_2, "test_2")]:
    for col_name in ["Multibranch_predict - train 1", "Multibranch_predict - train 2"]:
        header = col_name.split("- ")[-1] + " " + table_name
        score_printer(header, col_name, table)
        print("--------------------------")
```
